```python
#Exercise 1: For a program to prompt the user for hours and rate per hour to compute gross pay. 
#Rewrite your pay computation to give the employee 1.5 times the hourly rate for hours worked above 40 hours

total_hours = float (input('enter work hours:'))
rate = float (input('enter rate per hour:'))

if total_hours <= 40:  
    overtime_hour = 0
    regular_pay = total_hours * rate
    print ('Regular pay:', regular_pay)
    
else:
    regular_pay = 40 * rate
    overtime_hour = total_hours - 40
    overtime_pay = overtime_hour * 1.5 * rate
    total_pay = regular_pay + overtime_pay
    print ('overtime pay:', overtime_pay)
    print ('Total pay:', total_pay)
```

```python
#Write a function to_celsius(f) that converts Fahrenheit to Celsius using the formula:
#C = ((F−32) × 5)/9, Then test it with 100°F (should give 37.77°C).
```

```python
def to_celsius(f):
    C = ((f-32) * 5)/9
    return C
    
print("Celsisu:", to_celsius(100))
```
